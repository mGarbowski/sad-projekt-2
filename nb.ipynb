# Wizualizacja rozkładu statystyki testowej


$$T | H_0 \sim \mathcal{N} \left( -\frac{A^2N}{2 \sigma^2}, \frac{A^2N}{\sigma^2} \right)$$

$$T | H_1 \sim \mathcal{N} \left( \frac{A^2N}{2 \sigma^2}, \frac{A^2N}{\sigma^2} \right)$$

$$t_c = F^{-1}_{T|H_0}(1-\alpha) = -\frac{A^2N}{2\sigma^2} + \frac{A\sqrt{N}}{\sigma} \Phi^{-1}(1-\alpha)$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from ipywidgets import interact, FloatSlider, IntSlider

def plot_distributions(A, N, sigma2, alpha):
    """
    Plot the probability density functions of T under H_0 and H_1
    with critical threshold t_c, false alarm region (α), and detection region (1-β)
    
    T|H_0 ~ N(-A²N/(2σ²), A²N/σ²)
    T|H_1 ~ N(A²N/(2σ²), A²N/σ²)
    
    t_c = -A²N/(2σ²) + (A√N/σ) Φ⁻¹(1-α)
    """
    sigma = np.sqrt(sigma2)
    
    # Parameters for distributions
    mu0 = -A**2 * N / (2 * sigma2)
    mu1 = A**2 * N / (2 * sigma2)
    var = A**2 * N / sigma2
    std = np.sqrt(var)
    
    # Calculate critical threshold t_c
    t_c = mu0 + (A * np.sqrt(N) / sigma) * norm.ppf(1 - alpha)
    
    # Calculate beta and power (1-beta)
    beta = norm.cdf(t_c, mu1, std)
    power = 1 - beta
    
    # Create x range
    x_min = mu0 - 4 * std
    x_max = mu1 + 4 * std
    x = np.linspace(x_min, x_max, 1000)
    
    # Calculate PDFs
    pdf_h0 = norm.pdf(x, mu0, std)
    pdf_h1 = norm.pdf(x, mu1, std)
    
    # Create figure
    fig, ax = plt.subplots(figsize=(12, 6))
    
    # Plot distributions
    ax.plot(x, pdf_h0, 'b-', linewidth=2, label=f'$T|H_0 \\sim \\mathcal{{N}}({mu0:.3f}, {var:.3f})$')
    ax.fill_between(x, pdf_h0, alpha=0.2, color='blue')
    
    ax.plot(x, pdf_h1, 'r-', linewidth=2, label=f'$T|H_1 \\sim \\mathcal{{N}}({mu1:.3f}, {var:.3f})$')
    ax.fill_between(x, pdf_h1, alpha=0.2, color='red')
    
    # Shade rejection region (false alarm region) under H_0
    x_reject = x[x >= t_c]
    pdf_h0_reject = norm.pdf(x_reject, mu0, std)
    ax.fill_between(x_reject, pdf_h0_reject, alpha=0.5, color='blue', label=f'Fałszywy alarm ($\\alpha = {alpha:.4f}$)')
    
    # Shade detection region (power) under H_1
    x_detect = x[x >= t_c]
    pdf_h1_detect = norm.pdf(x_detect, mu1, std)
    ax.fill_between(x_detect, pdf_h1_detect, alpha=0.5, color='red', label=f'Poprawna detekcja ($1-\\beta = {power:.4f}$)')
    
    # Draw critical threshold line
    ax.axvline(t_c, color='green', linestyle='--', linewidth=2.5, label=f'$t_c = {t_c:.3f}$')
    
    # Display parameters and key values
    info_text = f'$A = {A:.3f}$, $N = {N}$, $\\sigma^2 = {sigma2:.3f}$, $\\alpha = {alpha:.4f}$\n'
    info_text += f'$t_c = {t_c:.3f}$, $\\beta = {beta:.4f}$, $1-\\beta = {power:.4f}$'
    ax.text(0.02, 0.98, info_text, 
            transform=ax.transAxes, fontsize=11, verticalalignment='top',
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.7))
    
    ax.set_xlabel('$T$', fontsize=12)
    ax.set_ylabel('Probability Density', fontsize=12)
    ax.set_title('Rozkłady statystyki testowej $T$', fontsize=14)
    ax.legend(fontsize=10, loc='upper right')
    ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()

# Create interactive plot with sliders
interact(
    plot_distributions,
    A=FloatSlider(value=1.0, min=0.1, max=3.0, step=0.1, description='A (amplituda)'),
    N=IntSlider(value=20, min=10, max=200, step=10, description='N (długość sygnału)'),
    sigma2=FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1, description='σ² (wariancja szumu)'),
    alpha=FloatSlider(value=0.01, min=0.01, max=0.99, step=0.001, description='α (poziom istotności)')
)
```
